```python
# Import libraries
import pandas as pd
from sklearn.ensemble import GradientBoostingClassifier

# Create the dataset
data = {
    'Student': ['S1', 'S2', 'S3', 'S4'],
    'CGPA': [6.0, 7.0, 8.0, 9.0],
    'Placement': [0, 0, 1, 1]
}

df = pd.DataFrame(data)

# Display dataset
print("Dataset:")
print(df)
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# -----------------------------
# Dataset
# -----------------------------
X = np.array([[6.0],
              [7.0],
              [8.0],
              [9.0]])

y = np.array([0, 0, 1, 1])

# -----------------------------
# Initial prediction
# Gradient Boosting starts
# with the mean of y
# -----------------------------
initial_prediction = np.mean(y)

pred = np.full(len(y), initial_prediction)

print("Initial Prediction:", initial_prediction)

# =====================================================
# ITERATION 1
# =====================================================

# Residual = Actual - Current Prediction
residual_1 = y - pred

print("\n--- Iteration 1 ---")
print("Residual values:", residual_1)

# Train a regression tree on residuals
tree1 = DecisionTreeRegressor(max_depth=1, random_state=42)
tree1.fit(X, residual_1)

# Tree prediction
tree_prediction_1 = tree1.predict(X)

print("Tree 1 prediction:", tree_prediction_1)

# Learning rate
learning_rate = 0.5

# New prediction
new_prediction_1 = pred + learning_rate * tree_prediction_1

print("New predicted values after Tree 1:", new_prediction_1)

# =====================================================
# ITERATION 2
# =====================================================

# New residual
residual_2 = y - new_prediction_1

print("\n--- Iteration 2 ---")
print("Residual values:", residual_2)

# Train second tree
tree2 = DecisionTreeRegressor(max_depth=1, random_state=42)
tree2.fit(X, residual_2)

# Second tree prediction
tree_prediction_2 = tree2.predict(X)

print("Tree 2 prediction:", tree_prediction_2)

# Final prediction
final_prediction = new_prediction_1 + learning_rate * tree_prediction_2

print("Final predicted values after Tree 2:", final_prediction)

# =====================================================
# Display everything in a table
# =====================================================

print("\n=================================================")
print("Student | CGPA | Actual | Residual1 | Tree1 | "
      "Pred1 | Residual2 | Tree2 | Final")
print("=================================================")

students = ["S1", "S2", "S3", "S4"]

for i in range(4):
    print(
        f"{students[i]:7} | "
        f"{X[i][0]:4.1f} | "
        f"{y[i]:6.1f} | "
        f"{residual_1[i]:9.2f} | "
        f"{tree_prediction_1[i]:5.2f} | "
        f"{new_prediction_1[i]:5.2f} | "
        f"{residual_2[i]:9.2f} | "
        f"{tree_prediction_2[i]:5.2f} | "
        f"{final_prediction[i]:5.2f}"
    )
```
